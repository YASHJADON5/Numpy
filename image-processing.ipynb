```python
from PIL import Image
import numpy as np
```

```python
img = Image.open("img1.png").convert("L")
print(img)
# print(img.shape)
print(img.size)
print(img.height)
print(img.width)
```

```python
image_array = np.array(img)
print(image_array)
```

```python
print(image_array.shape)
print(image_array.shape[0])
print(image_array.shape[1])
print(image_array.dtype)
```

```python
bright_image = image_array + 50
dark_image = np.clip(image_array -50,0,255) # 0 and 255 are the min and max range

Image.fromarray(bright_image).save("bright_image.png") #save
Image.fromarray(dark_image).show() #show
```

```python
increase_contrast = np.clip(128+1.1*(image_array-128),0,255)
Image.fromarray(increase_contrast).show()

decrease_contrast = np.clip(128+0.9*(image_array-128),0,255)
Image.fromarray(decrease_contrast).show()
```

```python
image_inversion = np.clip(255-image_array, 0, 255)
Image.fromarray(image_inversion).show()
```

```python
# convert the image into pure black and white image using thresholding
threshold = 128

image_bw = np.where(image_array >threshold, 255, 0).astype("uint8")

print(image_bw.dtype)

Image.fromarray(image_bw).show()
```
