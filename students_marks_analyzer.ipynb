```python
import numpy as np
```

```python
marks_arr= np.array([78, 56, 89, 45, 92, 67, 38, 81, 59, 73])
marks_arr
```

```python
names = np.array([
    "Yash",
    "Aman",
    "Riya",
    "Priya",
    "Rahul",
    "Neha",
    "Karan",
    "Anjali",
    "Mohit",
    "Sneha"
])
```

```python
highest_Marks = np.max(marks_arr)
index=np.argmax(marks_arr)
name_of_topper=names[index]
print(name_of_topper)
int(highest_Marks)
```

```python
lowest_Marks = np.min(marks_arr)
index=np.argmin(marks_arr)
name_of_lowest=names[index]
print(name_of_lowest)
int(lowest_Marks)
```

```python
average_Marks = np.mean(marks_arr)
int(average_Marks)
```

```python
median_Marks = np.median(marks_arr)
int(median_Marks)
```

```python
students_above_40=marks_arr[marks_arr>40].size
index_Of_Passed = np.where(marks_arr > 40)[0]
for index in index_Of_Passed:
    print(names[index], marks_arr[index])
```

```python
students_below_40=marks_arr[marks_arr<40].size
index_Of_Failed = np.where(marks_arr < 40)[0]
for index in index_Of_Failed:
    print(f"{names[index]}:{marks_arr[index]}")
```

```python
topper = np.argmax(marks_arr)
topper
```

```python
students_scored_above_average=np.where(marks_arr>average_Marks)[0]
students_scored_above_average
```
