```python
import numpy as np
```

```python

cities = np.array(["Delhi", "Mumbai", "Bangalore"])

temperature = np.array([
    [36, 37, 38, 35, 34, 36, 37],  # Delhi
    [30, 31, 32, 31, 30, 29, 30],  # Mumbai
    [25, 24, 26, 27, 25, 24, 26]   # Bangalore
])

days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
```

```python
for city, temp in zip(cities, temperature):
    print(f"Temperatures for {city} : {temp}")
```

```python
average_temp= np.array([])
for i in range(len(temperature)):
    average_temp  = np.append(average_temp,np.mean(temperature[i]))
    
for city,average in zip(cities, average_temp):
    print(f"Average temperature for {city} : {average}")
```

```python
average_temp_day_wise = np.mean(temperature, axis=0)

for day, average in zip(days, average_temp_day_wise):
    print(f"Average temperature for {day} : {average:.2f}")
```

```python
overall_avg_temp=np.mean(temperature)

print(f"Overall average temperature for the week : {overall_avg_temp:.2f}")
```

```python
r,c= np.where(temperature==np.max(temperature))
print(r,c)
print(temperature.shape)
print(f"The maximum temperature is of : {cities[r][0]} {temperature[r,c][0]}" )
```

```python
r,c= np.where(temperature == np.min(temperature))
print(r,c)
print(f"The coldest day is in {cities[r][0]} : {temperature[r,c][0]}")
```

```python
highest_Temperature_Of_Each_City = np.max(temperature,axis=1)
for city,highest_Temperature in zip(cities,highest_Temperature_Of_Each_City):

      print(f"Highest temperature of {city} : {highest_Temperature}")
```

```python
lowest_Temperature_Of_Each_City = np.min(temperature,axis=1)

for city,lowest_Temperature in zip(cities,lowest_Temperature_Of_Each_City ):

      print(f"Lowest temperature of {city} : {lowest_Temperature }")
```

```python
hottest_Day_Of_Each_City = np.argmax(temperature,axis=1)
print(hottest_Day_Of_Each_City)

for city,day in zip(cities,hottest_Day_Of_Each_City):
      print(f"Hottest day of {city} : {days[day]}")
```

```python
coldest_Day_Of_Each_City = np.argmin(temperature,axis=1)

for city,day in zip(cities,coldest_Day_Of_Each_City):
      print(f"Coldest day of {city} : {days[day]}")
```

```python
highest_Temperature = np.max(temperature)
coldest_Temperature = np.min(temperature)

print(f"The highest temperature of the week is : {highest_Temperature}")
print(f"The coldest temperature of the week is : {coldest_Temperature}")
```

```python
average_Temperature_Above_30 = np.mean(temperature,axis=1)


for city, avg_Temp in zip(cities,average_Temperature_Above_30):

      if(avg_Temp>30):

               print(f"{city} city has above temperature than 30")
```

```python
avg= np.mean(temperature,axis=1)
sortes_indices = np.argsort(avg)

for index in sortes_indices:
    print(cities[index], avg[index])
```
