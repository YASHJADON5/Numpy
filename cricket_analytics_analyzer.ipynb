```python
import numpy as np
```

```python
players = np.array([
    "Virat",
    "Rohit",
    "Gill",
    "Rahul",
    "Pant"
])

runs = np.array([78, 45, 102, 67, 89])

balls = np.array([56, 39, 90, 58, 61])
```

```python
index_of_highest = np.argmax(runs)
print("Highest Scorer")
print(f"{players[index_of_highest]} : {runs[index_of_highest]}")
```

```python
index_of_lowest = np.argmin(runs)
print("Lowest Scorer")
print(f"{players[index_of_lowest]} : {runs[index_of_lowest]}")
```

```python
strike_rate = np.array((runs / balls) * 100)

print("Strike Rate of all players:\n")



for player, strike_rate in zip(players, strike_rate):
    print(f"{player}: {strike_rate:.2f}")
```

```python
average_runs = np.array(runs/len(runs))
print(average_runs)
```

```python
average_runs = np.mean(runs)
print(average_runs)
```

```python
players_above_average = np.where(runs>average_runs)[0]
print("Players with above average score")
for index in players_above_average:
    print(f"{players[index]}: {runs[index]}")
```

```python
sort_by_runs = np.sort(runs)[::-1]
print(sort_by_runs)
```

```python
best_strike_rate = np.max(strike_rate)
print(best_strike_rate)
```

```python
fifties_And_Centuries =np.where((runs>=50) & (runs<=100))[0].size
print(fifties_And_Centuries)
```
