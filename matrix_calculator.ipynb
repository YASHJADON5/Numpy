```python
import numpy as np
```

```python
A = np.arange(1,10).reshape(3,3)
B= np.arange(10,19).reshape(3,3)
B[2,2]=50
A[2,2]=100
```

```python
print(A)
print(B)
```

```python
result_Sum_Matrix = np.add(A,B)
print(result_Matrix)
```

```python
result_Differenece_Matrix = np.subtract(B,A)
print(result_Differenece_Matrix)
```

```python
result_Dot_Matrix = np.dot(B,A)
print(result_Dot_Matrix)
```

```python
transpose_A = np.transpose(A)
print(transpose_A)
transpose_B= np.transpose(B)
print(transpose_B)
```

```python
determenant_Of_A= np.linalg.det(A)
print(determenant_Of_A)
determinant_Of_B= np.linalg.det(B)
print(determinant_Of_B)
```

```python
inverse_Of_A = np.linalg.inv(A)

inverse_Of_B = np.linalg.inv(B)

print(f"inverse of A is : {inverse_Of_A} and inverse of B is : {inverse_Of_B}")
```

```python
trace_Of_A= np.trace(A)
trace_Of_B= np.trace(B)
print(f"trace of A is : {trace_Of_A} and trace of B is : {trace_Of_B}")
```

```python
rank_Of_A= np.linalg.matrix_rank(A)
rank_Of_B= np.linalg.matrix_rank(B)
print(f"rank of A is : {rank_Of_A} and rank of B is : {rank_Of_B}")
```

```python
A_i= np.eye(A.shape[0])
res= np.array_equal(A, A_i)
res1 = np.array_equal(B,A_i)
print(res,res1)
```

```python
arr1 = np.array([1,2,3])
arr2= np.array([1,2,3])

res= np.array_equal(arr1,arr2)
print(res)
```

```python
print((A==np.eye(A.shape[0])).all())
print((B==np.eye(B.shape[0])).all())
```

```python
res = np.allclose(A, np.eye(A.shape[0]))
print(res)
res1 = np.allclose(B, np.eye(B.shape[0]))
print(res1)
```

```python
isSymmetric_A= (A == np.transpose(A)).all()
print(isSymmetric_A)
isSymmetric_B= (B == np.transpose(B)).all()
print(isSymmetric_B)
```
